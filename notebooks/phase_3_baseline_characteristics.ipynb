# Phase 3: baseline characteristics by documented sepsis status

This notebook compares adult hematologic-malignancy inpatient discharges with versus without documented sepsis (`A41*`). Results are discharge-level and survey-weighted with `DISCWT`. Choose **Run → Run All Cells** to refresh every output.

```python
from pathlib import Path
import json
import os
import sys
import pandas as pd
from IPython.display import Markdown, display

repo_root = Path.cwd()
if not (repo_root / 'src').is_dir():
    repo_root = repo_root.parent
if not (repo_root / 'src/phase_3.py').is_file():
    raise RuntimeError('Open this notebook from the IPC-MPC-Study repository.')
os.chdir(repo_root)
sys.path.insert(0, str(repo_root))
from src.phase_3 import main
summary = main()
output_dir = repo_root / 'outputs/phase_3'
```

## Definitions and interpretation

- **Population:** adult inpatient discharges with a draft qualifying HM diagnosis in any diagnosis position.
- **Exposure groups:** documented sepsis (`A41*`) versus no `A41*` code.
- **Weighted n:** cumulative national discharge estimate across 2016–2022, not unique patients.
- **Standardized difference:** descriptive group imbalance; positive values indicate a higher value or proportion in the sepsis group. Absolute values around 0.10 or greater are commonly treated as potentially meaningful imbalance.
- **Hospital variables:** decoded from the four-digit `NIS_STRATUM`.

```python
cohort_summary = pd.DataFrame([
    {'group': 'No documented sepsis', 'unweighted_n': summary['no_sepsis_unweighted_n'], 'weighted_n_2016_2022': summary['no_sepsis_weighted_n']},
    {'group': 'Documented sepsis (A41*)', 'unweighted_n': summary['sepsis_unweighted_n'], 'weighted_n_2016_2022': summary['sepsis_weighted_n']},
])
display(Markdown('## Cohort sizes'))
display(cohort_summary)
print('Copy/paste CSV:\n' + cohort_summary.to_csv(index=False))
```
