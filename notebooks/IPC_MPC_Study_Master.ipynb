# IPC-MPC Study: Complete Analysis Notebook

This is the primary, consolidated study notebook. Choose **Run → Run All Cells** to refresh the complete analysis in one place. It uses the cached adult hematologic-malignancy cohort and then runs every validated report-facing analysis from baseline characteristics through annual trends.

All results describe NIS hospitalizations rather than unique longitudinal patients. Report-facing tables include p-values and total or cohort-denominator rows.

```python
from pathlib import Path
import json, os, sys
import pandas as pd
from IPython.display import Markdown, display

REPO_ROOT = Path.cwd()
if REPO_ROOT.name == 'notebooks':
    REPO_ROOT = REPO_ROOT.parent
if not (REPO_ROOT / 'src').is_dir():
    raise RuntimeError('Open this notebook from the IPC-MPC-Study repository.')
os.chdir(REPO_ROOT)
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
COHORT_DB = REPO_ROOT / 'outputs/phase_1_2/hm_cohort.duckdb'
if not COHORT_DB.exists():
    raise RuntimeError('The cached HM cohort is missing. Run src.phase_1_2 first.')
print(f'Repository: {REPO_ROOT}')
print(f'Cached cohort: {COHORT_DB}')
```

# Phase 1–2: Data Readiness and Adult HM Cohort

The validated aggregate files are loaded directly here. This avoids the two malformed pasted-output cells in the older Phase 1–2 notebook while preserving its actual cohort results.

```python
phase12 = REPO_ROOT / 'outputs/phase_1_2'
phase12_tables = [
    ('Data-readiness audit', 'data_readiness_audit.csv'),
    ('Cohort overview', 'review_table_1_cohort_overview.csv'),
    ('Cohort by year', 'cohort_by_year.csv'),
    ('Mutually exclusive first-listed HM subtype', 'review_table_3_subtypes.csv'),
    ('HM subtype overlap', 'cohort_overlap.csv'),
    ('Hospital region', 'hospital_by_region.csv'),
    ('Hospital location/teaching', 'hospital_by_location_teaching.csv'),
    ('Hospital bed size', 'hospital_by_bed_size.csv'),
]
for title, filename in phase12_tables:
    display(Markdown(f'## {title}'))
    table = pd.read_csv(phase12 / filename, keep_default_na=False)
    display(table.style.hide(axis='index'))
    print('Copy/paste CSV:\n' + table.to_csv(index=False))
```
