# Complications by documented sepsis status

This analysis compares adult HM discharges with documented `A41*` sepsis against adult HM discharges without `A41*`. Diagnosis codes establish co-documentation during the hospitalization, not temporal order or causation.

```python
from pathlib import Path
import os
import sys
import pandas as pd
from IPython.display import Markdown, display

repo_root = Path.cwd()
if not (repo_root / 'src').is_dir():
    repo_root = repo_root.parent
os.chdir(repo_root)
sys.path.insert(0, str(repo_root))
from src.phase_4_complications import main
summary = main()
output_dir = repo_root / 'outputs/phase_4'
```

## Coding notes

- Severe sepsis includes `R65.2*`; septic shock is separately identified by `R65.21`.
- Atrial fibrillation/flutter includes `I48*` but excludes chronic AF (`I48.2*`).
- Pulmonary embolism and acute lower-extremity DVT use the exact codes specified in the protocol.
- Categories can overlap; percentages should not be summed.
- Weighted counts are cumulative national discharge estimates for 2016–2022, not unique patients.
- P-values are approximate two-sided two-proportion tests using unweighted sampled discharge counts. They do not account for hospital clustering. Benjamini–Hochberg FDR-adjusted p-values address multiple testing across the displayed complications.

```python
complications = pd.read_csv(output_dir / 'complications_by_sepsis.csv')
complications.loc[len(complications)] = {'complication': 'Total cohort denominator', 'no_sepsis_unweighted_n': summary['no_sepsis_unweighted_n'], 'no_sepsis_weighted_n': '4,182,853', 'no_sepsis_weighted_percent': '100.00', 'sepsis_unweighted_n': summary['sepsis_unweighted_n'], 'sepsis_weighted_n': '792,105', 'sepsis_weighted_percent': '100.00', 'absolute_difference_percentage_points': '—', 'prevalence_ratio': '—', 'p_value_unadjusted': '—', 'p_value_fdr_adjusted': '—', 'fdr_significant_at_0_05': '—'}
display(Markdown('## Table. Co-documented complications by sepsis status'))
display(complications)
print('Copy/paste CSV:\n' + complications.to_csv(index=False))
```

```python
largest = pd.DataFrame(summary['largest_absolute_differences'])[[
    'complication', 'no_sepsis_weighted_percent',
    'sepsis_weighted_percent', 'absolute_difference_percentage_points'
]]
display(Markdown('## Largest absolute prevalence differences'))
largest.loc[len(largest)] = {'complication': 'Total cohort denominator', 'no_sepsis_weighted_percent': '100.00', 'sepsis_weighted_percent': '100.00', 'absolute_difference_percentage_points': '—'}
display(largest)
```

```python
top = summary['largest_absolute_differences'][0]
report_text = f"""## Copy/paste interpretation

Among adult hematologic-malignancy inpatient discharges, the largest absolute difference among the evaluated co-documented complications was observed for {top['complication'].lower()}: {top['sepsis_weighted_percent']:.2f}% among discharges with documented sepsis versus {top['no_sepsis_weighted_percent']:.2f}% among discharges without documented sepsis, an absolute difference of {top['absolute_difference_percentage_points']:.2f} percentage points. These diagnosis-based findings do not establish whether a complication preceded or followed sepsis.
"""
display(Markdown(report_text))
print(report_text)
```
