# Commands 20A–20C: Annual Trends in Documented Palliative-Care Use

This notebook calculates annual weighted utilization and modified-Poisson estimated annual percent changes (EAPCs). Run all cells to refresh the analysis.

## Methods

EAPC is calculated as `100 × (exp(year coefficient) − 1)` from a DISCWT-weighted modified-Poisson model with robust year-specific `NIS_STRATUM` linearization. Command 20B tests the year-by-sepsis term; Command 20C jointly tests eight year-by-subtype terms. Cells with 10 or fewer unweighted palliative-care events are suppressed. Because hospital clusters are unavailable, inference is a strata-adjusted approximation.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display
REPO_ROOT=Path.cwd()
if REPO_ROOT.name=='notebooks': REPO_ROOT=REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path: sys.path.insert(0,str(REPO_ROOT))
from src.phase_11_trends import main
summary=main()
```

## Command 20A: Overall annual trend

```python
overall=pd.read_csv(REPO_ROOT/'outputs/phase_11/annual_overall_palliative_care.csv',keep_default_na=False)
overall.columns=['Year','Unweighted sample n','Unweighted palliative-care n','Weighted hospitalizations','Weighted palliative-care n','Weighted palliative-care %','95% CI lower, %','95% CI upper, %']
display(overall.style.hide(axis='index').format(thousands=','))
e=summary['command_20a_eapc']; trend_20a=pd.DataFrame([{'Measure':'Overall EAPC','EAPC, %':e['eapc_percent'],'95% CI':f"{e['ci_95_lower_percent']:.2f}–{e['ci_95_upper_percent']:.2f}",'P-value':e['p_value']},{'Measure':'Total','EAPC, %':'—','95% CI':'—','P-value':'—'}]); display(trend_20a.style.hide(axis='index'))
```

## Command 20B: Trends by sepsis status

```python
annual_sepsis=pd.read_csv(REPO_ROOT/'outputs/phase_11/annual_palliative_care_by_sepsis.csv',keep_default_na=False)
annual_sepsis.columns=['Year','Sepsis status','Unweighted sample n','Unweighted palliative-care n','Weighted hospitalizations','Weighted palliative-care n','Weighted palliative-care %','95% CI lower, %','95% CI upper, %','Within-year p-value']
display(annual_sepsis.style.hide(axis='index').format(thousands=','))
b=summary['command_20b']; trend_20b=pd.DataFrame([{'Cohort':'No documented sepsis','EAPC, %':b['no_sepsis_eapc']['eapc_percent'],'95% CI':f"{b['no_sepsis_eapc']['ci_95_lower_percent']:.2f}–{b['no_sepsis_eapc']['ci_95_upper_percent']:.2f}",'Trend p-value':b['no_sepsis_eapc']['p_value'],'Year × sepsis p-value':b['year_by_sepsis_interaction_p_value']},{'Cohort':'Documented sepsis','EAPC, %':b['sepsis_eapc']['eapc_percent'],'95% CI':f"{b['sepsis_eapc']['ci_95_lower_percent']:.2f}–{b['sepsis_eapc']['ci_95_upper_percent']:.2f}",'Trend p-value':b['sepsis_eapc']['p_value'],'Year × sepsis p-value':''},{'Cohort':'Total','EAPC, %':'—','95% CI':'—','Trend p-value':'—','Year × sepsis p-value':b['year_by_sepsis_interaction_p_value']}]); display(trend_20b.style.hide(axis='index'))
```

## Command 20C: Annual utilization among sepsis admissions by HM subtype

```python
subtype_annual=pd.read_csv(REPO_ROOT/'outputs/phase_11/annual_sepsis_palliative_care_by_subtype.csv',keep_default_na=False)
display(subtype_annual.style.hide(axis='index').format(thousands=','))
subtype_cells=subtype_annual[subtype_annual['hm_subtype']!='Total'].copy(); subtype_matrix=subtype_cells.pivot(index='hm_subtype',columns='year',values='weighted_palliative_care_percent').reset_index()
total_rates=annual_sepsis[(annual_sepsis['Sepsis status']=='Documented sepsis') & (annual_sepsis['Year']!='Total, 2016–2022')]; subtype_matrix.loc[len(subtype_matrix)]=['Total',*total_rates['Weighted palliative-care %'].tolist()]
subtype_matrix.columns=['HM subtype','2016','2017','2018','2019','2020','2021','2022']; display(subtype_matrix.style.hide(axis='index'))
```

```python
subtype_trends=pd.read_csv(REPO_ROOT/'outputs/phase_11/sepsis_subtype_trend_tests.csv',keep_default_na=False)
subtype_trends.columns=['HM subtype','EAPC, %','95% CI lower, %','95% CI upper, %','Trend p-value','Overall year × subtype p-value']
display(subtype_trends.style.hide(axis='index'))
print(f"Suppressed subtype-year cells: {summary['command_20c']['suppressed_cells']}")
```

## Interpretation

Documented inpatient palliative-care use increased overall and within both sepsis strata. The rate of increase differed modestly by sepsis status and differed across HM subtypes among sepsis hospitalizations. No subtype-year estimates required small-cell suppression.
