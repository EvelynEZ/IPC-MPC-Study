# Command 15: Documented Inpatient Palliative-Care Use by HM Subtype

This notebook reports documented inpatient palliative-care use across mutually exclusive hematologic-malignancy subtypes overall and separately for hospitalizations with and without documented sepsis. Run all cells to refresh the results.

## Methods and legend

- `Z51.5` is normalized to `Z515` and searched in any diagnosis position.
- HM subtype is mutually exclusive. For overlapping subtype groups, the first qualifying HM diagnosis in diagnosis-field order determines the displayed subtype; `multiple_hm_groups` remains available for sensitivity analysis.
- Counts and percentages use `DISCWT`; confidence intervals and overall Wald tests use year-specific `NIS_STRATUM`. Because `HOSP_NIS` is unavailable by study decision, discharges are the variance units, so inference is a strata-adjusted approximation rather than full NIS survey inference.
- The p-value shown once in each table is the overall test that palliative-care prevalence is equal across all nine subtypes. It is not a lymphoma-specific test.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display

REPO_ROOT = Path.cwd()
if REPO_ROOT.name == 'notebooks':
    REPO_ROOT = REPO_ROOT.parent
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
from src.phase_7_subtype_palliative_care import main
summary = main()
```

```python
def load_table(filename):
    data = pd.read_csv(REPO_ROOT / 'outputs/phase_7' / filename, keep_default_na=False)
    data = data.drop(columns='cohort')
    data.columns = [
        'HM subtype', 'Unweighted sample n', 'Unweighted palliative-care n',
        'Weighted hospitalizations', 'Weighted palliative-care n',
        'Weighted palliative-care %', '95% CI lower, %', '95% CI upper, %',
        'Overall p-value'
    ]
    return data

tables = {
    'All adult HM hospitalizations': load_table('subtype_palliative_care_all_hm.csv'),
    'HM without documented sepsis': load_table('subtype_palliative_care_no_sepsis.csv'),
    'HM with documented sepsis': load_table('subtype_palliative_care_sepsis.csv'),
}
```

```python
for title, table in tables.items():
    display(pd.DataFrame({'Cohort': [title]}).style.hide(axis='index'))
    display(table.style.hide(axis='index').format(thousands=','))
```

## Copy/paste-friendly Markdown

The next cell prints all three tables in plain Markdown.

```python
def print_markdown(dataframe, title):
    print(f'## {title}\n')
    headers = list(dataframe.columns)
    print('| ' + ' | '.join(headers) + ' |')
    print('|' + '|'.join(['---'] * len(headers)) + '|')
    for row in dataframe.astype(str).itertuples(index=False, name=None):
        print('| ' + ' | '.join(value.replace('|', '\\|') for value in row) + ' |')
    print()
for title, table in tables.items():
    print_markdown(table, title)
```

## Interpretation

Documented inpatient palliative-care utilization differs across mutually exclusive HM subtypes overall and within both sepsis strata (all overall p<0.001). These are unadjusted descriptive associations; differences may reflect case mix, illness severity, prognosis, coding, and hospital factors.
